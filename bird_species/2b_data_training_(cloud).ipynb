```python
import os
from google.colab import files

print("🔍 Searching for output files...")

# The files we want to rescue
target_files = ['training_curves.png', 'gradcam_sample.png', 'resnet50_bird_classifier.pth']
found_files = []

# Walk through the entire Colab directory
for root, dirs, filenames in os.walk('/content'):
    for filename in filenames:
        if filename in target_files:
            filepath = os.path.join(root, filename)
            found_files.append(filepath)
            print(f"✅ Found: {filepath}")

# Download them automatically
if not found_files:
    print("⚠️ Could not find the files. Did the Colab runtime restart?")
else:
    print("\n📥 Initiating downloads...")
    for file in found_files:
        try:
            files.download(file)
        except Exception as e:
            print(f"Error downloading {file}: {e}")
```
